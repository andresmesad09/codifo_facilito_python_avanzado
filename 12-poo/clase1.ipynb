```python
class Person(object):
    def __init__(self, name: str, id_number: int):
        self.name = name
        self.id_number = id_number

    def display(self):
        print(f"{self.name=}, {self.id_number=}")

    def details(self):
        print(f"Name: {self.name} and id: {self.id_number}")

class Employee(Person):
    def __init__(self, name, id_number, salary, post):
        super().__init__(name, id_number)
        self.salary = salary
        self.post = post

    def details(self):
        print(f"Empleado {self.name} - {self.id_number}")

    
emp1 = Employee("Andres", 1092344, 60000, "Python Developer")
emp1.display()
emp1.details()
print(Employee.__mro__)
```

# Herencia Múltiple

```python
class ElectronicDevice:
    def __init__(self, brand):
        self.brand = brand

    def power_on(self):
        return f"{self.brand} device is powered on"
    
# Parent class 2
class Camera:
    def take_photo(self):
        return "Taking a photo"
    
class Smartphone(ElectronicDevice, Camera):
    def make_call(self, number):
        return f"calling {number} from {self.brand} smartphone"
    
```

```python
phone = Smartphone("Samsung")
print(phone.power_on())
print(phone.take_photo())
print(phone.make_call("1234"))
```

# Encapsulamiento

```python
class Base:
    class_attribute = "Hi"
    __hidden_class_attribute = "Private hi"

    def __init__(self, color):
        self._color = color

    def __private_mehotd(self):
        print("Do something")

    def public_method(self):
        self.__private_mehotd()

    def get_color(self):
        return self._color
```

```python
my_class = Base("blue")
```

```python
my_class.__hidden_class_attribute
```

```python
my_class.__private_mehotd()
```

```python
my_class.public_method()
```

```python
my_class.get_color()
```

# Abstracción

## Creando interfaces formales

```python
from abc import ABCMeta
```

```python
class RemoteControlAbs(metaclass=ABCMeta):
    @classmethod
    def next_channel(self):
        pass
    @classmethod
    def prev_channel(self):
        pass
    @classmethod
    def turn_up_volume(self):
        pass
    @classmethod
    def turn_down_volume(self):
        pass
```

```python
rc = RemoteControlAbs()
```

```python
class SamsungRemote(RemoteControlAbs):
    def next_channel(self):
        print("Samsung->next_channel")
    def prev_channel(self):
        print("Samsung->prev_channel")
    def turn_up_volume(self):
        print("Samsung->turn_up_volume")
    def turn_down_volume(self):
        print("Samsung->turn_down_volume")
```

```python
samsung_rc = SamsungRemote()
```

```python
samsung_rc.next_channel()
```

```python
class LGRemote(RemoteControlAbs):
    def next_channel(self):
        print("LG->next_channel")
    def prev_channel(self):
        print("LG->prev_channel")
    def turn_up_volume(self):
        print("LG->turn_up_volume")
```

```python
lg_rm = LGRemote()
```
